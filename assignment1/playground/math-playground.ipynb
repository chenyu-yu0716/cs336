```python
import sympy
```

```python
# 定义两个二维旋转矩阵，角度分别为 theta1 和 theta2
theta1, theta2 = sympy.symbols("theta1 theta2")

R1 = sympy.Matrix(
    [[sympy.cos(theta1), -sympy.sin(theta1)], [sympy.sin(theta1), sympy.cos(theta1)]]
)

R2 = sympy.Matrix(
    [[sympy.cos(theta2), -sympy.sin(theta2)], [sympy.sin(theta2), sympy.cos(theta2)]]
)

# 计算旋转矩阵的乘积
R = R1 * R2

# 展开化简结果，便于观察
R_simplified = sympy.simplify(R)

# 显示两个旋转矩阵和它们的乘积
sympy.pprint(R1)
print("×")
sympy.pprint(R2)
print("=")
sympy.pprint(R_simplified)
```
